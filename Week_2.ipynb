**What Is Linear Regression?**

Linear Regression is a supervised machine learning algorithm used to predict a numerical value.

It looks at the relationship between input variables and an output variable.

For example:

Input: Hours studied

Output: Exam score

The model learns from existing examples and then makes predictions for new data.

In simple terms:

Give the model historical examples, and it learns a relationship that can help predict a number.

Why Is It Called Regression?

In machine learning, regression refers to predicting a numerical value.

**Features and Target: A Quick Recap**

Let’s return to our student dataset.

Target (y) - The value we want to predict. Eg: Exam score

Notice something important. We are giving the model examples where we already know the answer.

That is what makes Linear Regression a supervised machine learning algorithm.

**1. The Model Starts With a Line**

Think of Linear Regression as trying to find a line that does a good job of predicting the exam scores.

It could start with a line that isn’t very good.

For example, for a student who studied 6 hours, the model might predict a score of 64. But the actual score is 70. The prediction is off by 6 points.

For another student, the prediction might be off by 2 points. For another, maybe 5 points.

These differences are called errors or residuals.

The idea is simple:

The closer our predictions are to the actual values, the better our line is.

But now we have another problem. There isn’t just one error. There are errors for every student. We need some way to summarize all of them.



2. **Loss: How Bad Is Our Line?**

This is where the idea of loss comes in. Loss gives us a single way to describe:

“How wrong is my model overall?”

Imagine we have two possible lines.

Line A

Prediction error: 2
Prediction error: 3
Prediction error: 1
Prediction error: 4
Line B

Prediction error: 8
Prediction error: 6
Prediction error: 10
Prediction error: 7
Clearly, Line A is doing a better job.

A loss function allows the model to turn all those individual errors into a measure of overall performance.



So we can think of it as:

Smaller loss = better line

The model’s goal is therefore to find a line that gives it the smallest possible loss.

But how exactly should we calculate that loss?

3. **Least Squares: Why Square the Errors?**

One common approach used in Linear Regression is called Least Squares.

The basic idea is very simple:

Take each error, square it, and then combine those squared errors.

But why square the errors?

Because errors can go in different directions.

Suppose one prediction is 5 points too high and another is 5 points too low.

If we simply added the errors:

+5 + (-5) = 0

It would look like there was no error. But that’s obviously not true. The model made two mistakes.

Squaring fixes this:

5 becomes 25
-5 also becomes 25
Now both errors count. There is another useful effect. Squaring makes large errors much more important.

For example:

Error of 2 → 4
Error of 5 → 25
Error of 10 → 100
So being very far away from the actual value gets penalized heavily.

That’s why least squares encourages the model to find a line that doesn’t make large mistakes.

In simple terms:

Least Squares tries to find the line with the smallest overall squared error.

4. **But How Does the Model Find That Line?**

We now know what the model wants:

Find the line with the smallest loss.

But there could be many possible lines. How does the model actually search through them?

This is where Gradient Descent comes in.

Imagine you’re standing somewhere on a mountain, and your goal is to reach the lowest point in the valley. But you can’t see. You can only feel the slope underneath your feet.

You could:

Check which direction goes downhill.
Take a small step in that direction.
Check again.
Take another step downhill.
Keep repeating.
Eventually, you get closer to the bottom.

Gradient Descent works with a similar idea. The model starts with a line. It calculates its loss. Then it asks:

“Which direction should I move this line to reduce the loss?”

It makes a small adjustment. Then it calculates the loss again. If the loss became smaller, the model is moving in the right direction.

It continues making adjustments until it reaches a point where the loss is as small as it can reasonably get.

5. **So What Is Actually Changing?**

This is an important point. The data doesn’t change. Our students still studied the same number of hours and received the same scores.

What changes is the line.

The model adjusts the line’s:

slope - how quickly the predicted score changes as study hours increase
intercept - where the line starts
Together, these determine where the line sits on the graph. Gradient Descent keeps adjusting them to improve the predictions.

You don’t need to worry about the equations behind these adjustments yet.

For now, remember the intuition:

The model keeps nudging the line in a direction that reduces its error.

6. **When Does It Stop?**

Eventually, the model reaches a point where making small adjustments doesn’t meaningfully reduce the loss anymore.

It has reached a minimum.

Think back to our mountain example.

You started somewhere on the mountain and kept walking downhill. Eventually, you reach the bottom. For Linear Regression, we’re looking for the equivalent:

The line where the loss is as small as possible. That becomes our best-fitting line.

It doesn’t necessarily pass through every single data point. Instead, it is the line that gives the best overall fit according to the chosen loss.

**The Big Idea**

You don’t need to memorize equations yet.

Just remember this story:



And this is an important shift in how we should think about machine learning.

The model isn’t “looking at the data and knowing the answer.” It is optimizing.

It makes a prediction, measures how wrong it is, and adjusts itself to do better.

That’s the basic idea behind how a machine learning model learns.

**MAE - How far off are we on average?**

One of the easiest metrics to understand is Mean Absolute Error, or MAE.

Imagine our model makes four predictions:

Actual: 84 → Predicted: 82 → Error: 2
Actual: 72 → Predicted: 75 → Error: 3
Actual: 91 → Predicted: 88 → Error: 3
Actual: 65 → Predicted: 68 → Error: 3

The absolute errors are:

2, 3, 3, 3

MAE simply asks:

“On average, how far away were my predictions?”

Here, the average error is:

2.75 points

So we can say:

Our model is off by about 2.75 points on average.

That’s very intuitive and generally:

Lower MAE = better predictions

MSE - What if large mistakes matter more?

Now imagine two models.

Model A - Most predictions are only a few points off.
Model B - Most predictions are close, but occasionally it makes a huge mistake.
Sometimes that large mistake matters much more than several small ones.

That’s where Mean Squared Error (MSE) becomes useful. Instead of simply looking at the size of the error, we square the error.

So:

Error = 2 → 4
Error = 5 → 25
Error = 10 → 100

Notice what happened?

A large error becomes much more significant.

That’s the key idea behind MSE:

Large mistakes are penalized more heavily.

Again, you don’t need to memorize the formula yet.

RMSE - Bring the error back to the original scale

There’s a small problem with MSE.

If our target is exam score in points, squaring the errors means the resulting number isn’t in the same unit anymore.

That’s why we often use Root Mean Squared Error (RMSE). RMSE essentially takes the square root of MSE, bringing the result back to the original scale.

So if RMSE is:

4.2

we can roughly interpret that as:

Our predictions are typically off by around 4.2 points.

This makes RMSE easier to interpret than MSE.

R² — How much of the pattern are we explaining?

Another popular metric for regression is R-squared (R²).

Instead of asking:

“How many points off am I?”

R² asks something closer to:

“How much of the variation in the target can my model explain?”

An R² closer to 1 generally indicates that the model explains the observed variation well.
An R² around 0 means the model isn’t explaining much more than a simple baseline.
And R² can even be negative when the model performs worse than that baseline.
We don’t need to dive into the mathematics yet.

The important thing is that different metrics answer different questions.

Choosing an evaluation metric is part of understanding the problem you’re solving.

But here’s where things get interesting…

Imagine we build two models.

Model A

Training performance: Good

New-data performance: Good

Model B

Training performance: Almost perfect

New-data performance: Terrible

At first, Model B looks amazing.

So what happened?

The model learned TOO much.

Suppose our model sees thousands of students. It might start learning genuine relationships such as:

More preparation generally helps performance.

But it might also start picking up random quirks in the training data. For example:

Students who studied exactly 6.5 hours on Tuesdays tend to score higher.

Maybe that’s true in our training data. But it might be completely meaningless in the real world.



The model has started learning noise rather than just useful patterns.

So we get very good training performance but Poor performance on new data. This is called overfitting.

Now let’s go in the opposite direction.

What if our model is too simple?

Suppose we try to predict everyone’s exam score with:

Prediction = 70

That’s it.

It doesn’t care about:

Hours studied
Attendance
Previous score
Anything else
It’s clearly missing important patterns.



This is called underfitting. The model hasn’t learned enough. So we get poor training performance and usually poor performance on new data

Think of it like this

Imagine trying to fit a line through our exam-score data.

**Underfitting**

The model is too simple.
It misses important patterns.
Good fit

The model captures the meaningful pattern.
Overfitting

The model tries to follow every little fluctuation in the data.
It starts fitting the noise.
The goal isn’t to create the most complicated model. It’s to create a model that captures the right patterns.

And this brings us to Bias and **Variance**

These two words sound much scarier than they really are. Let’s keep them intuitive.

**Bias:**

“Is my model making assumptions that are too simple?”

A high-bias model tends to be too simple and misses important relationships. That’s closely related to underfitting.

**Variance:**

“Is my model too sensitive to the particular data it was trained on?”

A high-variance model can change significantly depending on the training data.

It may fit the training examples extremely well but struggle with new data.That’s closely related to overfitting.

A simple way to remember it

High Bias - “I’m too simple. I can’t capture the pattern.” → Underfitting

High Variance - “I’m paying too much attention to the training data.” → Overfitting

And somewhere in between, we want:

Enough complexity to learn the real pattern, but not so much that we memorize the noise.

That’s the bias-variance tradeoff.



This is one of the fundamental ideas in Machine Learning.

And this is why “100% accuracy” isn’t automatically impressive

“Our model has 99.9% accuracy!”

Your next question shouldn’t necessarily be:

“Wow! How did you do that?”

It should be:

“On what data?”

And then:

“Which metric are you using?”

“What happens on new data?”

“What is the baseline?”

“Are we overfitting?”

This is the mindset we need when working with Machine Learning.

A model can get 100% on the training data and still be a bad model.

Why?

Because sometimes the model doesn’t really learn the underlying pattern. It learns the training data too well including its noise and quirks. That’s overfitting.

So today’s question is:

If we know our model is overfitting, what can we actually do about it?

Two important ideas help us here:

Regularization and Cross-Validation.

1. First, let’s go back to our exam example 🎓

Imagine we’re trying to predict a student’s exam score. Our model looks at 2-3 features and learns how they relate to the exam score. But imagine we give the model 20 different features.

Now we have:

Hours studied
Attendance
Previous score
Sleep hours
Number of breaks
Phone usage
Favorite subject
Day of the week
And many more…
Some of these might actually help. But some might just be noise.

A complex model might start assigning importance to almost everything.

It could basically say:

“This tiny detail helped me predict these training examples, so I am going to use it!”

And that’s where things can go wrong.

2. Regularization: Don’t Make the Model More Complicated Than Necessary

Regularization gives the model an additional instruction:

“Fit the data, but don’t become unnecessarily complicated.”

Remember that our model has weights for its features. A weight tells us how strongly a feature influences the prediction.

For example:

Hours studied → weight = 5.2

Attendance → weight = 2.1

Previous score → weight = 3.8

Random feature → weight = 15.7

That last weight should make us suspicious.

Why would a random feature have such a huge influence?

It might be helping the model fit the training data but not because it represents a real pattern.

Regularization adds a penalty for large/complex weights.

So now the model isn’t simply trying to:

“Make my predictions as accurate as possible.”

It’s trying to:

“Make accurate predictions while keeping the model simpler.”

That’s the basic idea behind regularization.

3. L1 vs L2 Regularization

You may hear two names quite often:

L1 Regularization - It encourages some weights to become exactly zero.



The model is effectively saying:

“These features aren’t helping enough. I can ignore them.” That’s why L1 regularization can also help with feature selection.

L2 Regularization

L2 usually doesn’t make weights exactly zero.

Instead, it shrinks them.



The features are still there, but their influence is controlled.

4. But How Much Regularization Should We Use?

Here’s another problem.

What if we regularize too little?

The model might still overfit.

What if we regularize too much?

Now the model might become too simple and fail to capture useful patterns.

Remember the bias–variance discussion from Day 5?

This connects directly to it. So we need to find a reasonable balance. And that’s where cross-validation becomes useful.

5. Cross-Validation: Don’t Trust One Data Split

Earlier, we talked about training, validation, and test data. But there’s a problem with a single validation set.

What if we simply got a lucky split?

Maybe the validation data happened to contain examples that were particularly easy for our model. Then our model could look better than it really is.

Cross-validation gives us a better way to evaluate this.

6. K-Fold Cross-Validation

Let’s say we have 100 training examples.

Instead of creating one fixed validation set, we divide the data into 5 groups, or folds. Then we train and validate multiple times.

And we continue until every fold has been used for validation. So each example gets a chance to be part of the validation data. Finally, we look at the performance across all 5 rounds.

Instead of asking:

“How did my model perform on this one validation split?”

we ask:

“How consistently does my model perform across different subsets of the data?”

That’s a much more useful question.

7. Where Do Hyperparameters Come In?

Now we have another term to introduce:

Hyperparameters. Don’t let the name scare you.

Remember:

Parameters - These are learned by the model from the data. For example: weights

Hyperparameters - These are settings we choose before/during model training.

For example: How much regularization should we use?

We can use cross-validation to compare these choices. Maybe our results look like:



We might choose λ = 1 because it performed best during cross-validation.

The process of finding good hyperparameter values is called hyperparameter tuning.

The Big Idea 🧠

A good machine learning model isn’t the one that memorizes the training data.It’s the one that learns patterns that are useful beyond the training data.

That’s why we have tools like:

Regularization → control model complexity
Cross-validation → test consistency across different splits
Hyperparameter tuning → find settings that generalize well
And ultimately:

The goal is to build a model that performs well on data it hasn’t seen yet.

So, what is Classification?

Imagine Walmart wants to predict:

Will this customer buy the product or not?

The answer isn’t a number like $25 or 72%.

It’s a category:

Buy ✅
Don’t Buy ❌

That’s classification.

In simple terms:

Classification is when a machine learning model predicts which category something belongs to.

But how does the model actually decide?

We might think the model simply says:

“This customer will buy.”

But internally, many classification models work with probabilities.

For example, the model might look at information about a customer:

Previous purchases
Products viewed
Time spent on the website
Discounts clicked
Number of previous orders
And produce something like:

Probability of buying = 0.82

In other words:

82% likely to buy.

But we still need to turn that probability into an actual decision. That’s where a threshold comes in.



What is a threshold?

Suppose we set our threshold at 0.5.

We can say:

If probability ≥ 0.5 → Buy
If probability < 0.5 → Don’t Buy

The model gives us a probability.

The threshold helps turn that probability into a decision.

And this threshold doesn’t always have to be 0.5. Depending on the problem, we might choose a different threshold.

Why does the threshold matter?

Imagine we’re building a model to detect fraudulent transactions.

Would we rather accidentally flag a legitimate transaction as fraud?
Or would we rather miss an actual fraudulent transaction?
The answer depends on the cost of each mistake.

For some problems, catching as many positive cases as possible is extremely important.
For others, we may want to be more careful before labeling something as positive.
This is why classification isn’t simply:

“The model got 90% accuracy, so we’re done.”

We need to understand what kind of mistakes the model is making.

A quick look at classification algorithms

There are many ways to build a classification model. Some common ones are:

🌳 Decision Trees
Think of them as a series of questions:

Is the customer a repeat buyer?
↓
Did they view the product?
↓
Did they use a discount?

Eventually, the tree reaches a decision.

🌲 Random Forest

Instead of relying on one decision tree, it combines predictions from many trees.

📈 Logistic Regression

Despite having “regression” in its name, logistic regression is commonly used for classification. It estimates the probability of belonging to a class.

🤖 Neural Networks

These can learn much more complex patterns and are widely used for tasks such as image, text, and speech classification.

We don’t need to understand all of these algorithms today.

The important thing is to understand the problem they’re trying to solve.

And now, a problem with Accuracy

Let’s go back to our Walmart example. Suppose we have 100 customers. 95 customers don’t buy anything. Only 5 customers make a purchase.

Now imagine our model simply predicts:

“Don’t Buy” for everyone.

What happens?

It correctly predicts 95 customers.

So:

Accuracy = 95%

Sounds impressive, right? But think about it.

The model identified zero of the 5 customers who actually bought something.

So while the accuracy is 95%, the model isn’t useful for our actual goal.

This is why, for classification problems, we often need metrics beyond accuracy.
