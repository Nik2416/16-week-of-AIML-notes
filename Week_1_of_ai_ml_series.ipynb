**What Do We Do When Information Is Missing?**

Imagine you’re running an online store.

Hundreds of customers leave reviews about the products they bought.


Now imagine you’re trying to analyze this customer data to understand what people like.

But you notice something. Some customers gave a rating but didn’t mention their age. Some entered their age but skipped their location. Some left a review but didn’t provide a rating at all.



This is called **missing data.**

And as we discussed yesterday, real-world data is rarely perfect.

But today, let’s go one step further.

What do we actually do with those missing values?

The answer is:

It depends.

And that’s an important lesson in AI and machine learning.

There is no magic rule that says:

“Whenever data is missing, always fill it with this value.”

First, we need to understand why the information is missing.

**Step one: Can we understand the reason?**

Suppose your website had a technical issue for two hours. During that time, customer ratings were not recorded.

Now you know why those ratings are missing. That’s useful information.

But imagine you see that 20% of customers are missing their age, and you have absolutely no idea why.

Did they skip it?


Did the system fail?



Was the question optional?



Was there a problem during data collection?



If we can’t understand the logic behind why the data is missing, we need to be careful about making assumptions.

Sometimes, the best option is simply to remove those data points from the model.

For example, if only a small number of rows are missing important information, removing those rows may be safer than inventing values.

Because a wrong guess can sometimes be worse than having less data.

Of course, we can’t always do this. If 40% of our data is missing, deleting all those rows could mean losing a huge part of the dataset.

So then we look at other approaches.

**Option 1: Fill it with the Mean**

Let’s say we’re analyzing how much customers spend.

We have:

$40, $50, $60, ?, $70

One simple approach is to calculate the mean, or average, and use it to fill the missing value. This is called **mean imputation.**

It can work reasonably well when the values are distributed fairly evenly. But there’s a catch.

Imagine these purchase amounts:

$40, $45, $50, $55, $2,000

That one customer who spent $2,000 dramatically changes the average.

And suddenly, the mean doesn’t represent a typical customer anymore.

**Option 2: Use the Median**

The median is the middle value when numbers are arranged from smallest to largest. Unlike the mean, it is less affected by extreme values.(OUTLIERS)

That’s why the median can be useful when your data contains unusual purchases, salaries, house prices, or other extreme values.

For example, one customer spending $2,000 shouldn’t necessarily convince us that the average customer spends hundreds of dollars.

Sometimes, the middle value tells a more realistic story.

**Option 3: Use the Mode For CATEGORICAL Value**

Now let’s say we’re missing a customer’s most frequently purchased category.

For example:

Skincare

Skincare


Makeup


?



Skincare

We can’t calculate an average here. These are categories, not numbers.

Instead, we can use the mode - the value that appears most often.

In this case:

Skincare

Mode is commonly used when working with categorical information.

But what if other information can help us?

Let’s go back to our online store. A customer’s annual spending is missing.

But we know:

How often they shop



How many products they buy



Their membership type


How long they’ve been a customer



Those pieces of information may help us make a better estimate.

This is where more advanced techniques, such as MICE, can be useful.


**MICE** stands for Multiple Imputation by Chained Equations.

Instead of filling every missing value with one simple average, MICE looks at relationships between different variables.

For example:

Customers who shop frequently and buy premium products may have different spending patterns than occasional shoppers.

The idea is to use the information available around a missing value to make a more informed estimate.

But here’s the important part. More advanced doesn’t always mean more correct.

There is no single formula.

The bigger lesson

When a value is missing, we are often tempted to ask:

What should I put there?

But the better question is:

Why is it missing in the first place?

Because every time we fill in a missing value, we’re making an assumption. And those assumptions eventually become part of what our model learns.

Good AI doesn’t start by blindly fixing the data.

It starts by understanding what the data is trying to tell us, even when some of it is missing.

**When Something Doesn’t Look Right**

So far, we’ve talked about different types of data and what happens when information is missing.

But messy data has another problem. Sometimes, nothing is missing. Everything is there. And yet…

**something just doesn’t look right.**

Imagine you have thousands of invoices that need to be processed.

Most invoices look fairly similar.

Invoice amounts are within a normal range.


Dates follow expected formats.


Each invoice contains a few line items.


Then one invoice appears with:

Amount: $1,000,000

while most invoices are between $500 and $5,000. Or perhaps a document that usually contains 2–3 pages suddenly has 300 pages.

Or an employee record shows:

Age: 250

Nothing is technically missing. But these values are unusual. In data, we call these outliers.

But Is an Outlier Always a Mistake?

Not necessarily.

That $1,000,000 invoice could be:

❌ A typing error.

Or it could be:

✅ A legitimate transaction from a major customer.

That 300-page document?

❌ It could be a system error where pages were duplicated.

Or:

✅ It could genuinely be a large annual report.

This is why we shouldn’t automatically delete unusual data.

First, we need to identify it. Then, we need to understand it.

**How Do We Find Something Unusual?**

For numerical data, two common methods are:

1. **Z-Score**

Z-score asks:

How far is this value from what is typical?

It considers the average and how spread out the data is. For example, if most invoices are around $2,000, an invoice worth $1,000,000 would likely have a very large Z-score.



A common rule is that values with a Z-score greater than 3 or less than -3 may deserve investigation.


2. **IQR (Interquartile Range)**

IQR takes a slightly different approach.

Instead of focusing on the average, it looks at the middle 50% of the data. It finds the typical range where most observations sit. Anything far outside that range gets flagged as potentially unusual.

Think of it like looking at a group photo. You don’t focus on the exact average height of everyone. You look at where most people are standing and notice if one person is standing on a ladder.



That’s essentially the intuition behind IQR.

Outliers in Unstructured Data

Outliers aren’t limited to numbers. Imagine you have 100,000 images collected from the internet. Most images are family photos, but your dataset also contains:

Hurricane images


Product photos


Screenshots


Duplicate images


Completely unrelated pictures


There may not be a column saying:

outlier = yes

Instead, we can use techniques such as clustering to find groups of similar items.

For example, K-means clustering can group images based on their extracted features.

You might discover clusters like:



If a small cluster is completely different from the dominant group, it deserves investigation. But again, being unusual doesn’t automatically mean it’s bad data.

A hurricane image might be:

Useful if we’re building a disaster-detection model
Irrelevant if we’re building a family-photo classifier
What Can We Do With These Outliers?

Once we’ve identified unusual or irrelevant data, we can:

Delete it



→ If it’s clearly unrelated to the task.

Domain filter it

→ Keep only data that belongs to the problem we’re solving.

Deduplicate it



→ Remove repeated copies of the same image, document, or record.

Keep it



→ If the unusual data represents something our model actually needs to learn.

This is especially important when building AI systems from large collections of web data, images, documents, or other unstructured sources.

An outlier isn’t defined only by being numerically extreme. Sometimes it’s simply data that doesn’t belong to the pattern or domain you’re trying to learn.

Before training the model, we need to divide the data into different groups. But here’s the important question:

Can we just randomly divide the data?

Sometimes yes. Sometimes absolutely not.

The way we split data depends on the type of data and the problem we’re solving. Today, let’s understand the most common approaches:

* Random Split
* Stratified Split
* Time-Based Split
* Group-Based Split
* Training, Validation & Test Split

1. **Random Split 🎲**

The simplest approach is to randomly distribute data into different sets. Suppose we have 100 customer records. We might randomly split them:

80 records → Training
10 records → Validation
10 records → Test
Each customer has a random chance of ending up in any pile.

This works well when your data is:

Large enough


Randomly collected



Not dependent on time


Fairly balanced


For example:

Predicting house prices using features like bedrooms, location, and square footage.

Randomly splitting the houses can work well.

But Random Splitting Has a Problem ⚠️

Imagine you’re building a model to detect a rare disease. Out of 10,000 patients:

9,900 → Don’t have the disease



100 → Have the disease



Only 1% of the data belongs to the positive class.

Now imagine we randomly split the data.

By chance, one of the smaller datasets might contain very few or even no positive examples. That creates a problem. The training and test datasets may no longer represent the real distribution properly.

This is where Stratified Splitting helps.

2. **Stratified Split 📊**

A stratified split makes sure that the important categories maintain approximately the same proportion across each dataset. Let’s use the disease example.

With a stratified split, each dataset maintains approximately the same distribution.



Instead of simply shuffling and hoping for the best, we intentionally preserve the proportions.

Important point:

Stratified doesn’t mean the data isn’t random. It is still randomly selected—but within each category.

3. **Time-Based Split ⏰**

Now let’s consider a completely different problem. Imagine we’re predicting:

Will a customer cancel their subscription next month?

We have customer data from:

* January
* February
* March
* April
* May
* June


Would it make sense to randomly mix all six months together?

Not always.

Because in the real world:

The past predicts the future.

We should train using older data and test using newer data.



Why Not Randomly Split Time Data?

Imagine you’re predicting tomorrow’s stock price.

If your training data accidentally contains information from December while your test data contains January, the model is indirectly learning from the future.

That’s unrealistic. This is another form of data leakage.

4. **Group-Based Split 👥**


Sometimes the problem isn’t about time or class distribution.

It’s about people or groups.

Imagine you’re building a model using medical images.

You have:

1,000 patients


Each patient has 10 images


That’s 10,000 images.

If we randomly split individual images, we might get:

Patient A Image 1 → Training


Patient A Image 2 → Training



Patient A Image 3 → Test ❌

The model has already seen very similar images from Patient A during training.

The test isn’t truly testing something new.

**The Solution: Group-Based Split**

We keep all records belonging to the same person in one dataset.



This is useful when data contains multiple records from:

The same customer

The same patient

The same employee

The same company

The same household

The goal is:

Don’t let information from the same group leak into both training and testing.

5. **Training, Validation & Test Sets**

Regardless of how we perform the split, we usually assign different purposes to the datasets.

🟢 Training Set - Used to teach the model.

The model sees:

Input


Correct answers


And learns patterns.

🟡 Validation Set - Used while developing the model.

We might compare Model A vs. Model B

Or change settings such as:

Number of trees



Learning rate



Number of layers



The validation set helps us decide which version is better?

🔵 Test Set - The final untouched dataset.

We use it at the end to answer:

How well does this model perform on genuinely unseen data?

One Important Thing: The Split Happens Before Learning 🚨

A common mistake is allowing information from the full dataset to influence preprocessing.

For example:

Imagine you calculate the average income using the entire dataset before splitting. That means your training process has indirectly used information from the test data.

The Big Idea 💡

There is no universal: “Just use 80/20 split.”

The right way to split data depends on how the data was created and how the model will be used in the real world.
